```python
import numpy as np

class CalculativeFoundation:
    def __init__(self, data_matrix):
        """
        Initializes with a student-subject score matrix 
        Shape: (num_students, num_subjects)
        """
        self.data_matrix = np.array(data_matrix, dtype=float)

    def vector_fundamentals(self, student_idx1, student_idx2, sub_indices_3d=None):
        """Part A: Vector Norms, Dot Products, Angles, Cross Products, Projections"""
        v1 = self.data_matrix[student_idx1]
        v2 = self.data_matrix[student_idx2]
        
        # Norms
        norm1_v1 = np.linalg.norm(v1, ord=1)
        norm2_v1 = np.linalg.norm(v1, ord=2)
        
        # Dot Product & Angle
        dot_prod = np.dot(v1, v2)
        norm_product = np.linalg.norm(v1) * np.linalg.norm(v2)
        cos_angle = dot_prod / norm_product if norm_product != 0 else 0
        angle_rad = np.arccos(np.clip(cos_angle, -1.0, 1.0))
        angle_deg = np.degrees(angle_rad)
        
        # Projection of v1 onto v2
        projection = (np.dot(v1, v2) / np.dot(v2, v2)) * v2 if np.dot(v2, v2) != 0 else np.zeros_like(v2)
        
        # Cross product for 3D selected subjects (if provided)
        cross_prod = None
        if sub_indices_3d is not None and len(sub_indices_3d) == 3:
            s1_3d = v1[sub_indices_3d]
            s2_3d = v2[sub_indices_3d]
            cross_prod = np.cross(s1_3d, s2_3d)

        return {
            "v1": v1, "v2": v2,
            "norm1_v1": norm1_v1, "norm2_v1": norm2_v1,
            "dot_product": dot_prod, "angle_degrees": angle_deg,
            "projection_v1_on_v2": projection, "cross_product_3d": cross_prod
        }

    def matrix_operations(self):
        """Part B: Matrix Addition, Multiplication, Transpose, Inverse, Determinant"""
        mat = self.data_matrix
        transpose_mat = mat.T
        
        # Square matrix operations (using mat.T @ mat or square submatrix if non-square)
        square_mat = mat @ mat.T if mat.shape[0] != mat.shape[1] else mat
        
        det = np.linalg.det(square_mat)
        try:
            inv = np.linalg.inv(square_mat)
        except np.linalg.LinAlgError:
            inv = None # Singular matrix check

        return {
            "original_matrix": mat,
            "transpose": transpose_mat,
            "determinant": det,
            "inverse": inv
        }

    def eigen_decomposition(self):
        """Part D: Compute Eigenvalues and Eigenvectors of the Covariance Matrix"""
        # Covariance matrix across features/subjects (rowvar=False)
        cov_matrix = np.cov(self.data_matrix, rowvar=False)
        eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
        
        # Sort eigenvalues and eigenvectors in descending order
        sorted_idx = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[sorted_idx]
        eigenvectors = eigenvectors[:, sorted_idx]

        return {
            "covariance_matrix": cov_matrix,
            "eigenvalues": eigenvalues,
            "eigenvectors": eigenvectors
        }
```

#### part A: Vector & Matrix Fundamentals

```python
import numpy as np

class VectorCalculator:
    def __init__(self, student_matrix):
        """
        Initialize with a matrix of student grades/features.
        Rows represent students, columns represent subjects.
        """
        self.matrix = np.array(student_matrix, dtype=float)
        
    def vector_fundamentals(self, student_idx1, student_idx2, sub_indices_3d):
        v1_full = self.matrix[student_idx1]
        v2_full = self.matrix[student_idx2]
        
        # 1. Norm-1 (Manhattan Norm): sum of absolute values
        norm1_v1 = float(np.sum(np.abs(v1_full)))
        
        # 2. Norm-2 (Euclidean Norm): square root of sum of squared values
        norm2_v1 = float(np.linalg.norm(v1_full))
        
        # 3. Dot Product between student vectors
        dot_product = float(np.dot(v1_full, v2_full))
        norm_v2 = float(np.linalg.norm(v2_full))
        
        # 4. Angle between vectors (in degrees)
        cos_theta = dot_product / (norm2_v1 * norm_v2 + 1e-10)
        cos_theta = np.clip(cos_theta, -1.0, 1.0)
        angle_degrees = float(np.degrees(np.arccos(cos_theta)))
        
        # 5. Vector Projection of Student 1 on Student 2
        if norm_v2**2 > 0:
            projection_v1_on_v2 = (dot_product / (norm_v2**2)) * v2_full
        else:
            projection_v1_on_v2 = np.zeros_like(v2_full)
            
        # 6. Cross Product (using 3 specified subject dimensions)
        v1_3d = v1_full[sub_indices_3d]
        v2_3d = v2_full[sub_indices_3d]
        cross_product_3d = np.cross(v1_3d, v2_3d)
        
        return {
            'norm1_v1': norm1_v1,
            'norm2_v1': norm2_v1,
            'dot_product': dot_product,
            'angle_degrees': angle_degrees,
            'projection_v1_on_v2': projection_v1_on_v2.tolist(),
            'cross_product_3d': cross_product_3d.tolist()
        }

# --- Example Usage ---
# Sample data: 3 students across 4 subjects (e.g., Math, Physics, Chemistry, English)
sample_student_data = [
    [85, 90, 78, 92],  # Student 0
    [70, 75, 80, 85],  # Student 1
    [88, 82, 95, 79]   # Student 2
]

calc = VectorCalculator(sample_student_data)

# Run your exact query
results_a = calc.vector_fundamentals(student_idx1=0, student_idx2=2, sub_indices_3d=[0, 1, 2])

print(f"Norm-1 (Manhattan) of Student 0: {results_a['norm1_v1']:.2f}")
print(f"Norm-2 (Euclidean) of Student 0: {results_a['norm2_v1']:.2f}")
print(f"Dot Product between Student 0 & Student 2: {results_a['dot_product']:.2f}")
print(f"Angle between vectors: {results_a['angle_degrees']:.2f} degrees")
print(f"Vector Projection of Student 0 on Student 2: {results_a['projection_v1_on_v2']}")
print(f"Cross Product (3D subjects): {results_a['cross_product_3d']}")
```
